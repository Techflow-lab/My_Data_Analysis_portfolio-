```python
# ================================================================
#  PYTHON BUILT‑INS FOR DATA ANALYSIS — COMPLETE NOTEBOOK 
#  Grouped by use case, with examples, target types, and mutation notes
# ================================================================


# ================================================================
# 1. DATA INSPECTION & SUMMARY
# ================================================================

# len() — count items (list, tuple, set, dict, string, array)
data = [10, 20, 30]
print(len(data))  # -> 3 (NEW OBJECT: integer)

# type() — inspect type
print(type(3.14))  # -> <class 'float'> (NEW OBJECT)

# isinstance() — check type
print(isinstance("hello", str))  # -> True (NEW OBJECT: boolean)

# max(), min() — find extremes
nums = [5, 9, 2]
print(max(nums))  # -> 9 (NEW OBJECT)
print(min(nums))  # -> 2 (NEW OBJECT)

# sum() — sum numeric iterable
print(sum(nums))  # -> 16 (NEW OBJECT)

# sorted() — returns NEW sorted list (does NOT mutate original)
print(sorted(nums))  # -> [2, 5, 9] (NEW OBJECT)
print(nums)          # -> [5, 9, 2] (unchanged)


# ================================================================
# 2. DATA CLEANING & TRANSFORMATION
# ================================================================

# map() — apply function to each element (returns NEW iterator)
data = ["1", "2", "3"]
print(list(map(int, data)))  # -> [1, 2, 3] (NEW OBJECT)

# filter() — keep items matching condition (NEW iterator)
nums = [1, 2, 3, 4]
print(list(filter(lambda x: x % 2 == 0, nums)))  # -> [2, 4] (NEW OBJECT)

# zip() — combine iterables (NEW iterator)
names = ["A", "B"]
scores = [90, 80]
print(list(zip(names, scores)))  # -> [('A', 90), ('B', 80)] (NEW OBJECT)

# enumerate() — add index (NEW iterator)
fruits = ["apple", "banana"]
print(list(enumerate(fruits)))  # -> [(0, 'apple'), (1, 'banana')] (NEW OBJECT)

# round() — round number (NEW object)
print(round(3.14159, 2))  # -> 3.14 (NEW OBJECT)

# type conversion: int(), float(), str()
print(int("10"))     # -> 10 (NEW OBJECT)
print(float("3.5"))  # -> 3.5 (NEW OBJECT)
print(str(123))      # -> "123" (NEW OBJECT)


# ================================================================
# 3. BOOLEAN FILTERING & LOGIC
# ================================================================

# all() — True if all elements are True
print(all([True, True, False]))  # -> False (NEW OBJECT)

# any() — True if at least one element is True
print(any([False, False, True]))  # -> True (NEW OBJECT)


# ================================================================
# 4. DATA STRUCTURE CREATION & CONVERSION
# ================================================================

# list(), tuple(), set(), dict() — always create NEW objects
print(list((1, 2, 3)))  # -> [1, 2, 3]
print(tuple([1, 2, 3]))  # -> (1, 2, 3)
print(set([1, 1, 2]))    # -> {1, 2}
print(dict([("a", 1), ("b", 2)]))  # -> {'a': 1, 'b': 2}

# range() — NEW range object
print(range(5))  # -> range(0, 5)


# ================================================================
# 5. FILE & DATA INPUT
# ================================================================

# open() — returns NEW file object
# (Do not run without a real file)
# f = open("data.txt")  # NEW OBJECT


# ================================================================
# 6. INTROSPECTION & DEBUGGING
# ================================================================

# dir() — list attributes/methods (NEW list)
print(dir("hello"))  # -> list of string methods

# help() — prints documentation (no return value)
# help(str)

# getattr() — get attribute (NEW object)
print(getattr("hello", "upper")())  # -> "HELLO"

# hasattr() — check attribute (NEW boolean)
print(hasattr("hello", "split"))  # -> True

# setattr(), delattr() — mutate object attributes
class X:
    pass

setattr(X, "value", 10)  # MUTATES class
print(X.value)           # -> 10


# ================================================================
# 7. ITERATION CONTROL
# ================================================================

# iter() — NEW iterator
it = iter([1, 2, 3])
print(next(it))  # -> 1 (NEW OBJECT each call)

# reversed() — NEW reverse iterator
print(list(reversed([1, 2, 3])))  # -> [3, 2, 1] (NEW OBJECT)


# ================================================================
# 8. NUMERIC UTILITIES
# ================================================================

# abs() — absolute value (NEW object)
print(abs(-5))  # -> 5

# pow() — exponentiation (NEW object)
print(pow(2, 3))  # -> 8

# round() — rounding (NEW object)
print(round(7.89))  # -> 8


# ================================================================
# 9. EXECUTION & EVALUATION (rare in data analysis)
# ================================================================

# eval() — evaluate expression (NEW object)
print(eval("3 + 5"))  # -> 8

# exec() — executes code (NO return, MUTATES namespace)
namespace = {}
exec("x = 10", namespace)
print(namespace["x"])  # -> 10


# ================================================================
# 10. BUILT‑IN METHODS (IN‑PLACE vs NEW OBJECT)
# ================================================================

# -------------------------
# LIST METHODS (mostly in‑place)
# -------------------------

lst = [3, 1, 2]

lst.append(4)   # IN‑PLACE
lst.extend([5]) # IN‑PLACE
lst.insert(0, 0) # IN‑PLACE
print(lst)  # -> [0, 3, 1, 2, 4, 5]

print(lst.pop())  # -> 5 (NEW OBJECT)
print(lst)        # -> [0, 3, 1, 2, 4]

lst.remove(3)  # IN‑PLACE
lst.sort()     # IN‑PLACE
lst.reverse()  # IN‑PLACE
print(lst)     # -> [4, 2, 1, 0]


# -------------------------
# STRING METHODS (always NEW objects)
# -------------------------

s = "  Hello World  "

print(s.strip())      # -> "Hello World" (NEW OBJECT)
print(s.lower())      # -> "  hello world  " (NEW OBJECT)
print(s.replace("World", "Python"))  # -> "  Hello Python  " (NEW OBJECT)
print(s.split())      # -> ["Hello", "World"] (NEW OBJECT)


# -------------------------
# DICT METHODS (mixed)
# -------------------------

d = {"a": 1, "b": 2}

d.update({"c": 3})  # IN‑PLACE
print(d)            # -> {'a': 1, 'b': 2, 'c': 3}

print(d.pop("a"))   # -> 1 (NEW OBJECT)
print(d)            # -> {'b': 2, 'c': 3}

print(d.items())    # -> dict_items([...]) (NEW view object)
print(d.keys())     # -> dict_keys([...]) (NEW view object)
```
