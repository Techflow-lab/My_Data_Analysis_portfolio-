```python
convert_from = input("Enter Starting Unit of Measurement(inches, feet, yards): ")

convert_to = input("Enter Unit of Measurement to Conver to(inches, feet, yards): ")

number_of_inches = input("Enter Starting Measurement in Inches: ")

number_of_feet = input("Enter Starting Measurement in Feet: ")

number_of_yards = input("Enter Starting Measurement in Yards: ")
```

```python
if convert_from == "inches" and convert_to == "feet":
    result = float(number_of_inches) / 12
    print(f"{number_of_inches} inches is equal to {result} feet.")
elif convert_from == "feet" and convert_to == "inches":
    result = float(number_of_feet) * 12
    print(f"{number_of_feet} feet is equal to {result} inches.")
elif convert_from == "yards" and convert_to == "inches":
    result = float(number_of_yards) * 36
    print(f"{number_of_yards} yards is equal to {result} inches.")
elif convert_from == "yards" and convert_to == "feet":
    result = float(number_of_yards) * 3
    print(f"{number_of_yards} yards is equal to {result} feet.")
else:
    print("Invalid unit conversion.")
    
```
