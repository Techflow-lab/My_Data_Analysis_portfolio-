```python
top_countries = df['job_country'].value_counts().head(6).index
df_top_countries_salary = df.pivot_table(index = 'job_country', columns = 'job_title_short', values = 'salary_year_avg', aggfunc = 'median')
df_top_countries_salary = df_top_countries_salary.loc[top_countries]
job_titles = ['Data Analyst', 'Data Engineer', 'Data Scientist', 'Machine Learning Engineer']
df_top_countries_salary = df_top_countries_salary[job_titles]
df_top_countries_salary.plot(kind = "bar")
plt.title('job_most_provided_countries_salary')
plt.xlabel('each_country_data_roles')
plt.ylabel('salary')
plt.xticks(rotation = 45, ha = "right")
df_top_countries_salary
```

```python

df.reset_index(inplace = True)
df.index.name = 'job_index'
df.index
```

```python
df_usa = df[df['job_country'] == 'United States'].copy()
df_usa.reset_index(drop = True, inplace = True)
df_usa.head(3)
```

```python
df.head(3)
```

```python
df_usa['job_posted_month'] = df_usa['job_posted_date'].dt.strftime('%B')
df_usa
```

```python
df_usa_pivot = df_usa.pivot_table(index = 'job_posted_month', columns = 'job_title_short', aggfunc = 'size')
df_usa_pivot
```

```python
df_usa_pivot.reset_index(inplace = True)
```

```python
df_usa_pivot['month_number'] = pd.to_datetime(df_usa_pivot["job_posted_month"], format = "%B").dt.month
df_usa_pivot.sort_values(by = 'month_number', inplace = True)
df_usa_pivot.set_index('job_posted_month', inplace = True)
df_usa_pivot.drop(columns = 'month_number', inplace = True)
df_usa_pivot
```

```python
df_usa_pivot.plot(kind='line')
```

```python
top3 = df_usa['job_title_short'].value_counts().head(3)
top3 = top3.index.to_list()
top3
```

```python
df_usa_pivot[top3].plot(kind = 'line')
```

```python
df_usa_software_pivot = pd.read_csv('https://lukeb.co/software_csv', index_col='job_posted_month')
```

```python
df_usa_merged=df_usa_pivot.merge(df_usa_software_pivot, on = 'job_posted_month')
df_usa_merged
```

```python
top5 = df_usa_merged.sum().sort_values(ascending = False).head(5).index.to_list()
top5
```

```python
df_usa_merged[top5].plot(kind='line')
```
