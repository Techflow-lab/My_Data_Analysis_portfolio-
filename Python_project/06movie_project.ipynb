```python
movies['profit'] = movies['revenue'] - movies['budget']
movie_genres = movies[['popularity','budget','revenue','original_title','runtime','genres','release_date','vote_count','vote_average','profit']]
movie_genres.head()
```

```python
movies['profit'] = movies['revenue'] - movies['budget']
movies.head()
```

```python

split = movies['genres'].str.split('|').explode()
```

```python
split = pd.DataFrame(split)
split.columns = ['genres_split']
```

```python

movie_genres = movie_genres.join(split)
```

```python
movie_genres.info()
```

```python
genres_count = movie_genres.groupby('genres_split').count()['original_title'].sort_values(ascending = False)
genres_count = pd.DataFrame(genres_count)
genres_count
```

```python
genres_count.plot(kind = 'pie', y = 'original_title', figsize=(10,10))
```

```python
genres_count.plot(kind = 'barh', y = 'original_title', figsize=(10,10))
```

```python
genres_rev = movie_genres.groupby('genres_split')['revenue'].mean().sort_values(ascending = False)
genres_rev = pd.DataFrame(genres_rev)
genres_rev = genres_rev.sort_values(by = 'revenue', ascending =  False)
genres_rev
```

```python
genres_rev.plot(kind = 'bar', color = 'Darkblue', figsize = (10, 10))
```
