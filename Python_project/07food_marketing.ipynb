```python
import pandas as pd
food = pd.read_csv(r"C:\vscode\new_data_analysis\Python_project\u_food_marketing.csv")
food
```

```python
food.drop_duplicates(keep = False, inplace = True)
food
```

```python
food['total_children'] = food['Kidhome'] + food['Teenhome']
food.head()
```

```python
pd.set_option('display.max_columns', None)
food.head()
```

```python
#         !!!!!!    Conditional replacement
food['marital_Divorced'] = food['marital_Divorced'].replace({1:5, 0:0})
food['marital_Divorced'].unique()
```

```python
food['marital_Married'] = food['marital_Married'].replace({1:4,0:0})
food['marital_Single'] = food['marital_Single'].replace({1:3,0:0})
food['marital_Together'] = food['marital_Together'].replace({1:2,0:0})
food['marital_Widow'] = food['marital_Widow'].replace({1:1,0:0})
food['marital_status'] = food[['marital_Divorced','marital_Married','marital_Single','marital_Together','marital_Widow']].sum(axis = 1)
```

```python
food
```

```python
food['marital_status_str'] = food['marital_status'].map({5:'Divorced', 4:'Married',3:'Single',2:'Together',1:'Widow'})
```

```python
food['marital_status_str']
```

```python
food.info()
```

```python
food['education_2n Cycle'] = food['education_2n Cycle'].replace({1:1,0:0})
food['education_Basic'] = food['education_Basic'].replace({1:2,0:0})
food['education_Graduation'] = food['education_Graduation'].replace({1:3,0:0})
food['education_Master'] = food['education_Master'].replace({1:4,0:0})
food['education_PhD'] = food['education_PhD'].replace({1:5,0:0})
food['education_level'] = food[['education_2n Cycle','education_Basic','education_Graduation','education_Master','education_PhD']].sum(axis = 1)
food['education_level']
```

```python
food['accepted_campaigns'] = food[['AcceptedCmp1','AcceptedCmp2','AcceptedCmp3','AcceptedCmp4','AcceptedCmp5']].sum(axis = 1)
food[food['accepted_campaigns'] != 0]
```

```python
# Use if-else type function np.select() to assign new catagorical value to divide data into groups in the new column.
# Analyse everthing within each custom group!!!!!!
import numpy as np
conditions = [
    (food['Age'] >= 23) & (food['Age'] <= 30), 
    (food['Age'] >= 31) & (food['Age'] <= 40), 
    (food['Age'] >= 41) & (food['Age'] <= 50),
    (food['Age'] >= 51) & (food['Age'] <= 60), 
    (food['Age'] >= 61) & (food['Age'] <= 70),
    (food['Age'] >= 71) & (food['Age'] <= 80),   
]
choices = ['23~30', '31~40', '41~50','51~60','61~70','71~80']

food['Age_group'] = np.select(conditions, choices, default='need a fix')
food[['Age','Age_group']].head()
food['Age_group'].unique()
```

```python
# I want a decent order!
age_order = food.sort_values(by = 'Age')['Age_group'].unique()
age_order
```

```python
# And then groupby('x_axis')['y_axis'].mean() as data point on this graph.
# Lastly use the sample standard error to draw a vertical line as interval expand.
sns.pointplot(data = food, x = 'Age_group', y = 'accepted_campaigns', order = age_order)
```

```python
# Analyse total engagement of each group percentage campared with whole sample!!!!!!
counts = food['Age_group'].value_counts()
percentage =  counts / len(food)
percentage = percentage.reset_index()
percentage.columns = ['Age_group','percentage']
percentage = percentage.sort_values(by = 'Age_group')
```

```python
import matplotlib.pyplot as plt
sns.barplot(data = percentage, x = 'Age_group', y = 'percentage', palette = 'deep')
plt.title('Percentage of Campaign Accepted per Age Group')
plt.show()
```

```python
major_age_groups_spent = food.groupby('Age_group')['MntTotal'].sum()
major_age_groups_spent = major_age_groups_spent.reset_index()
major_age_groups_spent
```

```python
sns.barplot(data = major_age_groups_spent, x = 'Age_group', y = 'MntTotal', palette = 'deep')
plt.title('Amount_Spent_per_Group')
plt.show()
```

```python
major_age_groups_spent_filtered = food[food['accepted_campaigns'] != 0]
major_age_groups_spent_filtered = major_age_groups_spent_filtered.groupby('Age_group')['MntTotal'].sum().reset_index().sort_values(by = 'Age_group')
major_age_groups_spent_filtered = major_age_groups_spent_filtered.merge(major_age_groups_spent, on = 'Age_group', how  = 'inner', suffixes = ('_with_campaign','_all'))
major_age_groups_spent_filtered
```

```python
major_age_groups_spent_filtered = major_age_groups_spent_filtered.melt(id_vars = 'Age_group', value_vars = ['MntTotal_with_campaign','MntTotal_all'], var_name = 'Type', value_name = 'Value')
sns.barplot(data = major_age_groups_spent_filtered, x = 'Age_group', y = 'Value', hue = 'Type')
plt.title('Amount_Spent_per_Group_Comparision')
plt.show()
```
